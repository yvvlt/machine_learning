```python
import csv

dates = []
avg_temps = []
max_temps = []
min_temps = []

f = open('seoul.csv', 'r', encoding='cp949')
reader = csv.reader(f)
header = next(reader)

for row in reader:
    if row[-1] != '':
        dates.append(row[0])
        avg_temps.append(float(row[2]))
        max_temps.append(float(row[3]))
        min_temps.append(float(row[4]))

f.close()

print(dates[:3])
print(avg_temps[:3])
print(len(dates))
```

```python
import numpy as np

avg_arr = np.array(avg_temps)
max_arr = np.array(max_temps)
min_arr = np.array(min_temps)

print(f"평균: {np.mean(avg_arr):.2f}℃")
print(f"표준편차: {np.std(avg_arr):.2f}")
print(f"최고: {np.max(max_arr)}℃")
print(f"최저: {np.min(min_arr)}℃")

print()

diff_arr = max_arr - min_arr

print(f"평균 일교차: {np.mean(diff_arr):.2f}℃")
print(f"일교차가 가장 큰 날: {dates[np.argmax(diff_arr)]} ({np.max(diff_arr)}℃)")
print(f"일교차가 가장 작은 날: {dates[np.argmin(diff_arr)]} ({np.min(diff_arr)}℃)")
```

```python
hot_mask = max_arr >= 33
hot_days = np.array(dates)[hot_mask]
hot_temps = max_arr[hot_mask]

print(f"폭염일수: {hot_mask.sum()}일")
for d, t in zip(hot_days, hot_temps):
    print(f"{d}: {t}℃")
```

```python
import pandas as pd

df = pd.DataFrame({
    '날짜': pd.to_datetime(dates),
    '평균기온': avg_temps,
    '최고기온': max_temps,
    '최저기온': min_temps
})

df.set_index('날짜', inplace=True)
df['일교차'] = df['최고기온'] - df['최저기온']

df.head()
```

```python
import matplotlib.pyplot as plt

plt.plot(df.index, df['평균기온'])
plt.xlabel("year")
plt.ylabel("average_temp")
plt.show()
```

```python
import csv

student_ids = []
names = []
korean = []
english = []
math = []
science = []
social = []

f = open('student_scores.csv', encoding='utf-8-sig')
data = csv.reader(f)

next(data)

for row in data:
    student_ids.append(row[0])
    names.append(row[1])
    korean.append(int(row[2]))
    english.append(int(row[3]))
    math.append(int(row[4]))
    science.append(int(row[5]))
    social.append(int(row[6]))

f.close()

print(student_ids)
print(names)
print(korean)
print(english)
print(math)
print(science)
print(social)
```

```python
import numpy as np

kor = np.array(korean)
eng = np.array(english)
mat = np.array(math)
sci = np.array(science)
soc = np.array(social)

name = np.array(names)

scores = np.array([kor, eng, mat, sci, soc])
subjects = ['국어', '영어', '수학', '과학', '사회']

for i in range(5):
    print(subjects[i])
    print("평균 :", np.mean(scores[i]))
    print("최고점 :", np.max(scores[i]))
    print("최저점 :", np.min(scores[i]))
    print("표준편차 :", np.std(scores[i]))
    print()

# 전체 과목 최고점
max_score = np.max(scores)
idx = np.where(scores == max_score)

print("전체 최고점 :", max_score)
print("과목 :", subjects[idx[0][0]])

# 수학 최고점, 최저점 학생
max_i = np.argmax(mat)
min_i = np.argmin(mat)

print("수학 최고점 학생 :", name[max_i], mat[max_i])
print("수학 최저점 학생 :", name[min_i], mat[min_i])
```

```python
# 학생별 총점과 평균
total = kor + eng + mat + sci + soc
avg = total / 5

print("학생별 총점")
print(total)

print("학생별 평균")
print(avg)

# 평균 기준 상위 5명
rank = np.argsort(avg)[::-1]

print("\n평균 상위 5명")

for i in rank[:5]:
    print(name[i], round(avg[i], 2))
```

```python
# 국어 80점 이상
a = kor >= 80

print("국어 80점 이상 학생")

for i in range(len(name[a])):
    print(name[a][i], kor[a][i])

# 평균 60점 미만
b = avg < 60

print("\n평균 60점 미만 학생 수 :", np.sum(b))
```

```python
import pandas as pd

df = pd.DataFrame({
    '학번': student_ids,
    '이름': names,
    '국어': kor,
    '영어': eng,
    '수학': mat,
    '과학': sci,
    '사회': soc,
    '총점': total,
    '평균': avg
})

print("전체 통계")
print(df.describe())

print("\n평균 상위 5명")
print(df.sort_values('평균', ascending=False).head(5))

# 등급 계산
grade = []

for a in avg:
    if a >= 90:
        grade.append('A')
    elif a >= 80:
        grade.append('B')
    elif a >= 70:
        grade.append('C')
    elif a >= 60:
        grade.append('D')
    else:
        grade.append('F')

df['등급'] = grade

print("\n등급 추가 결과")
print(df)
```

```python
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 1. 과목별 평균 막대그래프
subjects = ['국어', '영어', '수학', '과학', '사회']

means = [
    np.mean(kor),
    np.mean(eng),
    np.mean(mat),
    np.mean(sci),
    np.mean(soc)
]

plt.bar(subjects, means)
plt.title('과목별 평균 점수')
plt.xlabel('과목')
plt.ylabel('평균 점수')
plt.show()


# 2. 평균 점수 히스토그램
plt.hist(avg, bins=10)
plt.title('전체 평균 점수 분포')
plt.xlabel('평균 점수')
plt.ylabel('학생 수')
plt.show()


# 3. 수학과 과학 산점도
plt.scatter(mat, sci)
plt.title('수학 vs 과학 점수')
plt.xlabel('수학 점수')
plt.ylabel('과학 점수')
plt.show()


# 4. 등급별 파이차트
count = df['등급'].value_counts()

plt.pie(count, labels=count.index, autopct='%.1f%%')
plt.title('등급별 인원 비율')
plt.xlabel('등급별 인원 비율')
plt.ylabel('인원 비율')
plt.show()
```
